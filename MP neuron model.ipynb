Mp Neuron Model for OR boolean Function

```python
x1=int(input("Enter the value of x1:"))
x2=int(input("Enter the value of x2:"))
result=x2+x1
if(result>=1):
    print("Output of the neuron is 1")
else:
    print("Output of the neuron is 0")
```

Mp Neuron Model for AND boolean Function

```python
x1=int(input("Enter the value of x1:"))
x2=int(input("Enter the value of x2:"))
result=x2+x1
if(result>=2):
    print("Output of the neuron is 1")
else:
    print("Output of the neuron is 0")
```

```python
def mp_neuron(x1, x2, threshold):
    if x1 not in (0, 1) or x2 not in (0, 1):
        raise ValueError("Inputs must be 0 or 1")

    net = x1 + x2
    return 1 if net >= threshold else 0


# OR
print("OR Output:", mp_neuron(1, 0, threshold=1))

# AND
print("AND Output:", mp_neuron(1, 0, threshold=2))
```
